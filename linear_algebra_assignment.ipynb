### Q2 Part (iii) — NumPy Function

Write a NumPy function that:
- Takes **any 2×2 matrix** and **any 2D vector** as inputs
- Returns the **transformed vector**
- Prints a **one-line description** of how the x and y components individually change

Test your function on matrix `A` and all four vectors above.

````python
# Input — do not modify
A = np.array([[2, 0], [0, 3]])
v1 = np.array([1, 0])
v2 = np.array([0, 1])
v3 = np.array([1, 1])
v4 = np.array([-2, 4])
````

````python
def transform_and_describe(matrix, vector):
    """
    Multiplies a 2x2 matrix by a 2D vector, prints a description of
    the component changes, and returns the new vector.
    """
    # Perform matrix multiplication using the @ operator (or np.dot)
    transformed = matrix @ vector

    # Print a one-line description of how x and y individually changed
    print(f"x changed from {vector[0]} to {transformed[0]}, and y changed from {vector[1]} to {transformed[1]}")

    return transformed

vectors = {'v1': v1, 'v2': v2, 'v3': v3, 'v4': v4}

# Test the function on matrix A and all four vectors
for name, v in vectors.items():
    print(f"Testing {name}:")
    new_v = transform_and_describe(A, v)
    print(f"Result: {new_v}\n")
````

### Q3 Part (iii) — The Problem with Raw Dot Products

A senior engineer tells you:
> *"Raw dot products are a bad measure of similarity — you need to normalize."\*

1. What is the **problem** with raw dot products?
2. Construct a **specific example using NumPy** that demonstrates the flaw.
3. Fix it.

> **Hint:** Take a user with scores `[1.8, 0.2, 1.6, 0.4]`

The fundamental flaw with the raw dot product is that it measures two things at the same time: alignment (how similar the tastes are) and magnitude (how big the numbers are).
Because the math is just multiplying and adding, a user who simply rates everything highly (or watches way more movies) will generate massive dot products.

````python
# Write your code here:

#2. Demonstrating the Flaw with NumPy
user_a = np.array([0.9, 0.1, 0.8, 0.2])
# The Hint: Exact same taste ratios, but double the magnitude
user_e = np.array([1.8, 0.2, 1.6, 0.4])

# The Loud Mismatch: Different taste, but huge numbers
user_loud = np.array([3.0, 3.0, 0.0, 0.0])

print(f"A's similarity to themselves: {np.dot(user_a, user_a):.2f}")
print(f"A's similarity to User E:     {np.dot(user_a, user_e):.2f}")
print(f"A's similarity to User Loud:  {np.dot(user_a, user_loud):.2f}")

#3. The Fix: Cosine Similarity (Normalization)

def cosine_similarity(v1, v2):
    # Divide the dot product by the product of both magnitudes
    dot_product = np.dot(v1, v2)
    mag1 = np.linalg.norm(v1)
    mag2 = np.linalg.norm(v2)
    return dot_product / (mag1 * mag2)

print(f"Fixed A & A:    {cosine_similarity(user_a, user_a):.2f}")
print(f"Fixed A & E:    {cosine_similarity(user_a, user_e):.2f}")
print(f"Fixed A & Loud: {cosine_similarity(user_a, user_loud):.2f}")
````

### Q3 Part (iv) — Most Similar Users

After normalizing, which **two users are most similar**?

Does that match your intuition from just reading the vectors?

What does it mean in plain language — what do these two users have in common?

> **Note:** A similarity score is just a number that tells you **how alike two things are**.

User  A and C most similar with cosine similarity score of 0.99(out of perfect 1.0).

It matches with intuition as If you look at the pattern, both users rate the first and third genres (Action and Thriller) extremely high, while giving very low scores to the second and fourth genres (Romance and Comedy). Their numbers are just slightly swapped around, but the overall "shape" of their taste is almost identical.

In plain English, a high cosine similarity score means these two people are going to enjoy the exact same movie recommendations.Even though User A might like Action slightly more than Thrillers, and User C likes Thrillers slightly more than Action, they both aggressively prefer intense, high-stakes movies over lighthearted or romantic ones.

````python
# Write your code here:
user_names = ['A', 'B', 'C', 'D']
magnitudes = np.linalg.norm(users, axis=1, keepdims=True)
normalized_users = users / magnitudes
cosine_sim_matrix = normalized_users @ normalized_users.T

print("--- Full Cosine Similarity Matrix ---")
print(np.round(cosine_sim_matrix, 2))
print("\n")

search_matrix = cosine_sim_matrix.copy()
np.fill_diagonal(search_matrix, -1) # Ignore self-matches

# Locate the highest score
flat_index = np.argmax(search_matrix)
row_idx, col_idx = np.unravel_index(flat_index, search_matrix.shape)

user_1 = user_names[row_idx]
user_2 = user_names[col_idx]
max_score = search_matrix[row_idx, col_idx]

print("--- Most Similar Pair ---")
print(f"Users {user_1} and {user_2} with a score of {max_score:.4f}")
````
